# Library Book Data Analyzer
### PDS Micro Project

This Jupyter Notebook analyzes the `library_books.csv` file using Pandas, NumPy, Matplotlib and Seaborn.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
sns.set_theme(style="whitegrid")
```

```python
# Load CSV dataset
file_path = "library_books.csv"
df = pd.read_csv(file_path)

print("LIBRARY BOOK DATA ANALYZER")
print("=" * 50)
print("Dataset loaded successfully.")
print("Total records:", len(df))
print("Columns:", list(df.columns))
```

```python
# Display all books
df
```

## Dataset Information

```python
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
```

## Library Statistics

```python
print("LIBRARY STATISTICS")
print("=" * 50)
print("Total Books:", len(df))
print("Total Authors:", df["Author"].nunique())
print("Total Genres:", df["Genre"].nunique())
print("Average Rating:", round(df["Rating"].mean(), 2))
print("Total Available Copies:", df["Available_Copies"].sum())
print("Average Book Price:", round(df["Price"].mean(), 2))

print("\nGenre Count:")
print(df["Genre"].value_counts())
```

## Rating and Price Analysis

```python
highest_rating = df["Rating"].max()
lowest_rating = df["Rating"].min()
highest_price = df["Price"].max()
lowest_price = df["Price"].min()

print("Highest Rating:", highest_rating)
print("Lowest Rating:", lowest_rating)
print("Most Expensive Price:", highest_price)
print("Cheapest Price:", lowest_price)

print("\nHighest Rated Books:")
display(df.loc[df["Rating"] == highest_rating,
              ["Book_Name", "Author", "Rating"]])

print("Most Expensive Books:")
display(df.loc[df["Price"] == highest_price,
              ["Book_Name", "Price"]])
```

## Advanced Analysis

```python
print("Average Rating by Genre:")
display(df.groupby("Genre")["Rating"].mean()
        .sort_values(ascending=False)
        .round(2)
        .to_frame("Average_Rating"))

print("Available Copies by Genre:")
display(df.groupby("Genre")["Available_Copies"].sum()
        .sort_values(ascending=False)
        .to_frame("Available_Copies"))

print("Publication Year Count:")
display(df["Publication_Year"].value_counts()
        .sort_index()
        .to_frame("Book_Count"))

print("Top 10 Authors:")
display(df["Author"].value_counts().head(10).to_frame("Book_Count"))
```

## NumPy Analysis

```python
ratings = np.array(df["Rating"], dtype=float)
prices = np.array(df["Price"], dtype=float)

print("RATING ANALYSIS USING NUMPY")
print("Maximum Rating:", np.max(ratings))
print("Minimum Rating:", np.min(ratings))
print("Average Rating:", round(np.mean(ratings), 2))

print("\nPRICE ANALYSIS USING NUMPY")
print("Maximum Price:", np.max(prices))
print("Minimum Price:", np.min(prices))
print("Average Price:", round(np.mean(prices), 2))
```

## Search, Filter and Sort

```python
def search_by_id(book_id):
    result = df[df["Book_ID"].astype(str) == str(book_id)]
    if result.empty:
        print("No book found.")
    else:
        display(result)

def search_by_name(name):
    result = df[df["Book_Name"].astype(str)
                .str.contains(name, case=False, na=False)]
    if result.empty:
        print("No book found.")
    else:
        display(result)

def filter_by_rating(min_rating):
    return df[df["Rating"] >= min_rating].sort_values(
        "Rating", ascending=False
    )

def sort_books(column):
    return df.sort_values(column)

print("Books with rating >= 4:")
display(filter_by_rating(4))
```

## Add, Update and Delete

```python
def add_book(book_id, book_name, author, genre, publication_year,
             rating, available_copies, price, language):
    global df

    if str(book_id) in df["Book_ID"].astype(str).values:
        raise ValueError("Book ID already exists.")
    if not 1900 <= int(publication_year) <= 2100:
        raise ValueError("Invalid publication year.")
    if not 0 <= float(rating) <= 5:
        raise ValueError("Rating must be between 0 and 5.")
    if int(available_copies) < 0:
        raise ValueError("Available copies cannot be negative.")
    if float(price) < 0:
        raise ValueError("Price cannot be negative.")

    new_book = pd.DataFrame([{
        "Book_ID": book_id,
        "Book_Name": book_name,
        "Author": author,
        "Genre": genre,
        "Publication_Year": int(publication_year),
        "Rating": float(rating),
        "Available_Copies": int(available_copies),
        "Price": float(price),
        "Language": language
    }])

    df = pd.concat([df, new_book], ignore_index=True)
    df.to_csv(file_path, index=False)
    return new_book

def update_book(book_id, **changes):
    global df
    mask = df["Book_ID"].astype(str) == str(book_id)

    if not mask.any():
        raise ValueError("Book not found.")

    for column, value in changes.items():
        if column not in df.columns or column == "Book_ID":
            raise ValueError(f"Invalid column: {column}")
        df.loc[mask, column] = value

    df.to_csv(file_path, index=False)
    return df.loc[mask]

def delete_book(book_id):
    global df
    mask = df["Book_ID"].astype(str) == str(book_id)

    if not mask.any():
        raise ValueError("Book not found.")

    deleted = df.loc[mask].copy()
    df = df.loc[~mask].reset_index(drop=True)
    df.to_csv(file_path, index=False)
    return deleted

print("Add, update and delete functions are ready.")
```
